## Test løsningen din

Sjekker at `f` og `g` gir riktige verdier i noen kjente punkter. Dette sier ingenting om selve tegningen — bare om funksjonsuttrykkene stemmer.


```python
def test_f_og_g():
    feil = []
    for x, forventet in [(-5, 23), (-1, 7), (0, 3), (3, -9)]:
        if f(x) != forventet:
            feil.append(f"f({x}) ga {f(x)!r}, forventet {forventet!r}")
    for x, forventet in [(-5, -11), (-2, 4), (0, 4), (1, 1), (3, -11)]:
        if g(x) != forventet:
            feil.append(f"g({x}) ga {g(x)!r}, forventet {forventet!r}")
    if feil:
        print("\n".join(feil))
    else:
        print("f og g ser riktige ut!")

test_f_og_g()
```
